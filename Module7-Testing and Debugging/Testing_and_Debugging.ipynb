```python
import pandas as pd

df = pd.read_excel(
    "military_final.xlsx",
    sheet_name="Wide_Format"
)

df.columns = df.columns.str.strip()

print("Dataset loaded successfully!")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
```

```python
print("Dataset Shape:", df.shape)

print("\nColumn Names:")
for i, col in enumerate(df.columns, start=1):
    print(f"{i}. {col}")
```

```python
# Cell 6 - Duplicate Country Check

duplicate_count = df["country"].duplicated().sum()

print("Duplicate country records:", duplicate_count)

if duplicate_count == 0:
    print("PASS: No duplicate countries found.")
else:
    print("FAIL: Duplicate countries found.")

    print("\nDuplicate records:")
    display(df[df["country"].duplicated(keep=False)].sort_values("country"))
```

```python
# Cell 7 - Missing Value Check

missing_values = df.isnull().sum()

total_missing = missing_values.sum()
missing_percentage = (total_missing / df.size) * 100

print("Total missing values:", total_missing)
print("Overall missing percentage:", round(missing_percentage, 2), "%")

if missing_percentage < 2:
    print("PASS: Missing values are below the 2% project target.")
else:
    print("FAIL: Missing values exceed the 2% project target.")

print("\nColumns with missing values:")

missing_table = pd.DataFrame({
    "Column": missing_values.index,
    "Missing Values": missing_values.values
})

missing_table = missing_table[
    missing_table["Missing Values"] > 0
]

if missing_table.empty:
    print("No missing values found.")
else:
    display(missing_table)
```

```python
# Cell 8 - Required Dashboard Columns Check

required_columns = [
    "country",
    "power_index",
    "power_index_rank",
    "power_index_rank_gap",
    "gdp",
    "gdp_rank",
    "defense_budget_usd",
    "budget_to_gdp_ratio",
    "total_population",
    "total_assets",
    "assets_per_capita",
    "active_personnel",
    "reserve_personnel",
    "paramilitary",
    "total_military_aircraft",
    "total_military_helicopters",
    "tanks",
    "armored_fighting_vehicles",
    "total_naval_fleet",
    "fighter_aircraft",
    "attack_aircraft",
    "transport_aircraft",
    "trainer_aircraft",
    "special_mission_aircraft",
    "tanker_aircraft",
    "attack_helicopters",
    "continent",
    "region",
    "alliance",
    "nato_flag"
]

missing_columns = [
    col for col in required_columns
    if col not in df.columns
]

print("Required columns:", len(required_columns))
print("Missing columns:", len(missing_columns))

if len(missing_columns) == 0:
    print("\nPASS: All required dashboard columns are present.")
else:
    print("\nFAIL: Some required columns are missing.")
    print("\nMissing columns:")
    for col in missing_columns:
        print("-", col)
```

```python
# Cell 9 - Numeric Data Validation

numeric_columns = [
    "power_index",
    "power_index_rank",
    "power_index_rank_gap",
    "gdp",
    "gdp_rank",
    "defense_budget_usd",
    "budget_to_gdp_ratio",
    "total_population",
    "total_assets",
    "assets_per_capita",
    "active_personnel",
    "reserve_personnel",
    "paramilitary",
    "total_military_aircraft",
    "total_military_helicopters",
    "tanks",
    "armored_fighting_vehicles",
    "total_naval_fleet",
    "fighter_aircraft",
    "attack_aircraft",
    "transport_aircraft",
    "trainer_aircraft",
    "special_mission_aircraft",
    "tanker_aircraft",
    "attack_helicopters"
]

numeric_results = []

for column in numeric_columns:
    converted = pd.to_numeric(df[column], errors="coerce")
    invalid_count = converted.isna().sum()

    numeric_results.append({
        "Column": column,
        "Invalid Values": invalid_count,
        "Status": "PASS" if invalid_count == 0 else "FAIL"
    })

numeric_results_df = pd.DataFrame(numeric_results)

display(numeric_results_df)

failed = (numeric_results_df["Status"] == "FAIL").sum()

print("\nTotal columns tested:", len(numeric_results_df))
print("Passed:", len(numeric_results_df) - failed)
print("Failed:", failed)

if failed == 0:
    print("PASS: All tested numeric columns contain valid numeric values.")
else:
    print("FAIL: Some columns contain invalid numeric values.")
```

```python
# Cell 10 - Power Index Rank Gap KPI Validation
import numpy as np
expected_rank_gap = (
    pd.to_numeric(df["gdp_rank"], errors="coerce")
    - pd.to_numeric(df["power_index_rank"], errors="coerce")
)

actual_rank_gap = pd.to_numeric(
    df["power_index_rank_gap"],
    errors="coerce"
)

rank_gap_check = np.isclose(
    actual_rank_gap,
    expected_rank_gap,
    rtol=1e-6,
    atol=1e-6
)

print("POWER INDEX RANK GAP TEST")
print("=" * 50)

print("Total records tested:", len(df))
print("Correct values:", rank_gap_check.sum())
print("Incorrect values:", (~rank_gap_check).sum())

if rank_gap_check.all():
    print("\nPASS: Power Index Rank Gap is correctly calculated.")
else:
    print("\nFAIL: Power Index Rank Gap contains incorrect values.")
    
    print("\nIncorrect records:")
    display(
        df.loc[
            ~rank_gap_check,
            [
                "country",
                "gdp_rank",
                "power_index_rank",
                "power_index_rank_gap"
            ]
        ]
    )
```

```python
# Cell 11 - Assets per Capita KPI Validation

expected_assets_per_capita = (
    pd.to_numeric(df["total_assets"], errors="coerce")
    /
    pd.to_numeric(df["total_population"], errors="coerce")
)

actual_assets_per_capita = pd.to_numeric(
    df["assets_per_capita"],
    errors="coerce"
)

assets_check = np.isclose(
    actual_assets_per_capita,
    expected_assets_per_capita,
    rtol=1e-6,
    atol=1e-10
)

print("ASSETS PER CAPITA TEST")
print("=" * 50)

print("Total records tested:", len(df))
print("Correct values:", assets_check.sum())
print("Incorrect values:", (~assets_check).sum())

if assets_check.all():
    print("\nPASS: Assets per Capita is correctly calculated.")
else:
    print("\nFAIL: Assets per Capita contains incorrect values.")

    print("\nIncorrect records:")
    display(
        df.loc[
            ~assets_check,
            [
                "country",
                "total_assets",
                "total_population",
                "assets_per_capita"
            ]
        ]
    )
```

```python
# Cell 12 - Budget-to-GDP Ratio KPI Validation

expected_budget_ratio = (
    pd.to_numeric(df["defense_budget_usd"], errors="coerce")
    /
    pd.to_numeric(df["gdp"], errors="coerce")
)

actual_budget_ratio = pd.to_numeric(
    df["budget_to_gdp_ratio"],
    errors="coerce"
)

budget_ratio_check = np.isclose(
    actual_budget_ratio,
    expected_budget_ratio,
    rtol=1e-6,
    atol=1e-10
)

print("BUDGET-TO-GDP RATIO TEST")
print("=" * 50)

print("Total records tested:", len(df))
print("Correct values:", budget_ratio_check.sum())
print("Incorrect values:", (~budget_ratio_check).sum())

if budget_ratio_check.all():
    print("\nPASS: Budget-to-GDP Ratio is correctly calculated.")
else:
    print("\nFAIL: Budget-to-GDP Ratio contains incorrect values.")

    print("\nIncorrect records:")
    display(
        df.loc[
            ~budget_ratio_check,
            [
                "country",
                "defense_budget_usd",
                "gdp",
                "budget_to_gdp_ratio"
            ]
        ]
    )
```

```python
# Cell 13 - Region Filter Validation

print("REGION FILTER TEST")
print("=" * 50)

regions = df["region"].dropna().unique()

print("Number of regions:", len(regions))
print("Regions found:")
print(list(regions))

test_region = regions[0]

region_filtered = df[df["region"] == test_region]

print("\nTest region:", test_region)
print("Records returned:", len(region_filtered))

if len(region_filtered) > 0:
    print("PASS: Region filter is working.")
else:
    print("FAIL: Region filter returned no records.")
```

```python
# Cell 14 - Continent Filter Validation

print("CONTINENT FILTER TEST")
print("=" * 50)

continents = df["continent"].dropna().unique()

print("Number of continents:", len(continents))
print("Continents found:")
print(list(continents))

test_continent = continents[0]

continent_filtered = df[
    df["continent"] == test_continent
]

print("\nTest continent:", test_continent)
print("Records returned:", len(continent_filtered))

if len(continent_filtered) > 0:
    print("PASS: Continent filter is working.")
else:
    print("FAIL: Continent filter returned no records.")
```

```python
# Cell 15 - Alliance Filter Validation

print("ALLIANCE FILTER TEST")
print("=" * 50)

alliances = df["alliance"].dropna().unique()

print("Number of alliance categories:", len(alliances))
print("Alliance categories:")
print(list(alliances))

test_alliance = alliances[0]

alliance_filtered = df[
    df["alliance"] == test_alliance
]

print("\nTest alliance:", test_alliance)
print("Records returned:", len(alliance_filtered))

if len(alliance_filtered) > 0:
    print("PASS: Alliance filter is working.")
else:
    print("FAIL: Alliance filter returned no records.")
```

```python
# Cell 16 - Country Selection Validation

print("COUNTRY SELECTION TEST")
print("=" * 50)

countries = df["country"].dropna().unique()

print("Total countries:", len(countries))

test_country = countries[0]

country_data = df[
    df["country"] == test_country
]

print("Selected country:", test_country)
print("Records returned:", len(country_data))

if len(country_data) == 1:
    print("PASS: Country selection returns exactly one country.")
else:
    print("FAIL: Country selection returned unexpected records.")
```

```python
# Cell 17 - Nation Overview Validation

print("NATION OVERVIEW TEST")
print("=" * 50)

test_country = "India"

if test_country in df["country"].values:

    nation = df[
        df["country"] == test_country
    ].iloc[0]

    nation_metrics = {
        "Country": nation["country"],
        "Power Index Rank": nation["power_index_rank"],
        "Power Index": nation["power_index"],
        "Defense Budget": nation["defense_budget_usd"],
        "Active Personnel": nation["active_personnel"],
        "Reserve Personnel": nation["reserve_personnel"],
        "Military Aircraft": nation["total_military_aircraft"],
        "Military Helicopters": nation["total_military_helicopters"],
        "Tanks": nation["tanks"],
        "Naval Fleet": nation["total_naval_fleet"]
    }

    for metric, value in nation_metrics.items():
        print(f"{metric}: {value}")

    print("\nPASS: Nation Overview data retrieved successfully.")

else:
    print("FAIL: Test country was not found.")
```

```python
# Cell 18 - Compare Powers Validation

print("COMPARE POWERS TEST")
print("=" * 50)

country_a = "India"
country_b = "China"

if country_a in df["country"].values and country_b in df["country"].values:

    a = df[df["country"] == country_a].iloc[0]
    b = df[df["country"] == country_b].iloc[0]

    comparison = pd.DataFrame({
        "Metric": [
            "Power Index Rank",
            "Defense Budget",
            "Active Personnel",
            "Military Aircraft",
            "Naval Fleet"
        ],
        country_a: [
            a["power_index_rank"],
            a["defense_budget_usd"],
            a["active_personnel"],
            a["total_military_aircraft"],
            a["total_naval_fleet"]
        ],
        country_b: [
            b["power_index_rank"],
            b["defense_budget_usd"],
            b["active_personnel"],
            b["total_military_aircraft"],
            b["total_naval_fleet"]
        ]
    })

    display(comparison)

    print("PASS: Compare Powers data generated successfully.")

else:
    print("FAIL: One or both countries were not found.")
```

```python
# Cell 19 - Compare Powers Selection Validation

print("COMPARE POWERS SELECTION TEST")
print("=" * 50)

available_countries = set(df["country"].dropna())

test_cases = [
    ("India", "China"),
    ("United States", "Russia"),
    ("France", "Germany")
]

results = []

for country_a, country_b in test_cases:

    valid_a = country_a in available_countries
    valid_b = country_b in available_countries

    results.append({
        "Country 1": country_a,
        "Country 2": country_b,
        "Country 1 Found": valid_a,
        "Country 2 Found": valid_b,
        "Status": "PASS" if valid_a and valid_b else "FAIL"
    })

comparison_tests = pd.DataFrame(results)

display(comparison_tests)
```

```python
# Cell 20 - Coalition Builder Validation

print("COALITION BUILDER TEST")
print("=" * 50)

coalition_countries = [
    "India",
    "France",
    "Germany"
]

coalition_df = df[
    df["country"].isin(coalition_countries)
].copy()

print("Selected coalition countries:")
print(list(coalition_df["country"]))

print("\nCountries successfully selected:", len(coalition_df))

if len(coalition_df) == len(coalition_countries):
    print("PASS: All coalition countries were found.")
else:
    print("FAIL: Some coalition countries were not found.")
```

```python
# Cell 21 - Coalition Metrics Validation

coalition_budget = pd.to_numeric(
    coalition_df["defense_budget_usd"],
    errors="coerce"
).sum()

coalition_personnel = pd.to_numeric(
    coalition_df["active_personnel"],
    errors="coerce"
).sum()

coalition_aircraft = pd.to_numeric(
    coalition_df["total_military_aircraft"],
    errors="coerce"
).sum()

coalition_naval = pd.to_numeric(
    coalition_df["total_naval_fleet"],
    errors="coerce"
).sum()

coalition_results = pd.DataFrame({
    "Metric": [
        "Defense Budget",
        "Active Personnel",
        "Military Aircraft",
        "Naval Fleet"
    ],
    "Coalition Total": [
        coalition_budget,
        coalition_personnel,
        coalition_aircraft,
        coalition_naval
    ]
})

display(coalition_results)

if coalition_results["Coalition Total"].notna().all():
    print("PASS: Coalition metrics calculated successfully.")
else:
    print("FAIL: Coalition metrics contain invalid values.")
```

```python
# Cell 22 - Coalition vs Reference Country

reference_country = "China"

if reference_country in df["country"].values:

    reference = df[
        df["country"] == reference_country
    ].iloc[0]

    coalition_comparison = pd.DataFrame({
        "Metric": [
            "Defense Budget",
            "Active Personnel",
            "Military Aircraft",
            "Naval Fleet"
        ],
        "Coalition": [
            coalition_budget,
            coalition_personnel,
            coalition_aircraft,
            coalition_naval
        ],
        "Reference Country": [
            reference["defense_budget_usd"],
            reference["active_personnel"],
            reference["total_military_aircraft"],
            reference["total_naval_fleet"]
        ]
    })

    display(coalition_comparison)

    print("PASS: Coalition comparison generated successfully.")

else:
    print("FAIL: Reference country not found.")
```

```python
# Cell 23 - Top 10 Power Index Validation

top_10_power = (
    df[
        [
            "country",
            "power_index_rank",
            "power_index"
        ]
    ]
    .sort_values(
        "power_index_rank",
        ascending=True
    )
    .head(10)
)

print("TOP 10 COUNTRIES BY POWER INDEX RANK")
print("=" * 50)

display(top_10_power)

if len(top_10_power) == 10:
    print("PASS: Top 10 Power Index list generated successfully.")
else:
    print("FAIL: Top 10 list could not be generated.")
```

```python
# Cell 24 - Defense Budget Validation

top_10_budget = (
    df[
        [
            "country",
            "defense_budget_usd"
        ]
    ]
    .sort_values(
        "defense_budget_usd",
        ascending=False
    )
    .head(10)
)

print("TOP 10 COUNTRIES BY DEFENSE BUDGET")
print("=" * 50)

display(top_10_budget)

if len(top_10_budget) == 10:
    print("PASS: Top 10 Defense Budget list generated successfully.")
else:
    print("FAIL: Defense Budget ranking failed.")
```

```python
# Cell 25 - Aircraft Data Validation

aircraft_columns = [
    "total_military_aircraft",
    "fighter_aircraft",
    "attack_aircraft",
    "transport_aircraft",
    "trainer_aircraft",
    "special_mission_aircraft",
    "tanker_aircraft"
]

aircraft_results = []

for column in aircraft_columns:

    values = pd.to_numeric(
        df[column],
        errors="coerce"
    )

    aircraft_results.append({
        "Metric": column,
        "Valid Records": values.notna().sum(),
        "Invalid Records": values.isna().sum(),
        "Negative Values": (values < 0).sum()
    })

aircraft_test = pd.DataFrame(aircraft_results)

display(aircraft_test)

if (
    aircraft_test["Invalid Records"].sum() == 0
    and aircraft_test["Negative Values"].sum() == 0
):
    print("PASS: Aircraft data validation completed successfully.")
else:
    print("FAIL: Aircraft data contains invalid values.")
```

```python
# Cell 26 - Personnel Data Validation

personnel_columns = [
    "active_personnel",
    "reserve_personnel",
    "paramilitary"
]

personnel_results = []

for column in personnel_columns:

    values = pd.to_numeric(
        df[column],
        errors="coerce"
    )

    personnel_results.append({
        "Metric": column,
        "Valid Records": values.notna().sum(),
        "Invalid Records": values.isna().sum(),
        "Negative Values": (values < 0).sum()
    })

personnel_test = pd.DataFrame(personnel_results)

display(personnel_test)

if (
    personnel_test["Invalid Records"].sum() == 0
    and personnel_test["Negative Values"].sum() == 0
):
    print("PASS: Personnel data validation completed successfully.")
else:
    print("FAIL: Personnel data contains invalid values.")
```

```python
# Cell 27 - Dashboard Module Validation

expected_modules = [
    "Quick Stats",
    "Nation Overview",
    "Compare Powers",
    "Coalition Builder"
]

print("EXPECTED DASHBOARD MODULES")
print("=" * 50)

for module in expected_modules:
    print("PASS:", module)

print("\nAll four dashboard modules are included in the project specification.")
```

```python
# Cell 28 - Final QA Summary

qa_results = []

def add_test(test_name, result):
    qa_results.append({
        "Test": test_name,
        "Status": "PASS" if result else "FAIL"
    })


# Dataset tests
add_test(
    "Dataset contains 140+ countries",
    len(df) >= 140
)

add_test(
    "No duplicate countries",
    df["country"].duplicated().sum() == 0
)

add_test(
    "Required dashboard columns present",
    len(missing_columns) == 0
)

# Missing values
add_test(
    "Missing values below 2%",
    (df.isnull().sum().sum() / df.size * 100) < 2
)

# KPI tests
add_test(
    "Power Index Rank Gap",
    rank_gap_check.all()
)

add_test(
    "Assets per Capita",
    assets_check.all()
)

add_test(
    "Budget-to-GDP Ratio",
    budget_ratio_check.all()
)

# Filter tests
add_test(
    "Region filter",
    len(region_filtered) > 0
)

add_test(
    "Continent filter",
    len(continent_filtered) > 0
)

add_test(
    "Alliance filter",
    len(alliance_filtered) > 0
)

add_test(
    "Country selection",
    len(country_data) == 1
)

# Dashboard feature tests
add_test(
    "Nation Overview",
    test_country in df["country"].values
)

add_test(
    "Compare Powers",
    country_a in df["country"].values
    and country_b in df["country"].values
)

add_test(
    "Coalition Builder",
    len(coalition_df) > 0
)

# Data validation
add_test(
    "Aircraft data validation",
    aircraft_test["Invalid Records"].sum() == 0
)

add_test(
    "Personnel data validation",
    personnel_test["Invalid Records"].sum() == 0
)


qa_summary = pd.DataFrame(qa_results)

display(qa_summary)

total_tests = len(qa_summary)
passed_tests = (qa_summary["Status"] == "PASS").sum()
failed_tests = (qa_summary["Status"] == "FAIL").sum()

print("=" * 50)
print("FINAL QA SUMMARY")
print("=" * 50)
print("Total Tests :", total_tests)
print("Passed      :", passed_tests)
print("Failed      :", failed_tests)

if failed_tests == 0:
    print("\nOVERALL STATUS: PASS")
else:
    print("\nOVERALL STATUS: REVIEW REQUIRED")
```

```python
# Cell 29 - Export QA Results

qa_output = "QA_Test_Results.xlsx"

qa_summary.to_excel(
    qa_output,
    index=False
)

print("QA results exported successfully!")
print("File:", qa_output)
```
