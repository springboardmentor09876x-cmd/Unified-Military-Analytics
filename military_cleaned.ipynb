```python
import pandas as pd

df = pd.read_csv("military_raw_data.csv")

print(df.head())
print(df.columns)
```

```python
import re

def clean_number(x):
    if pd.isna(x):
        return x

    x = str(x)

    x = x.replace(",", "")
    x = x.replace("%", "")
    x = x.replace("+", "")

    try:
        return float(x)
    except:
        return x

df = df.applymap(clean_number)
```

```python
df.to_csv("military_cleaned.csv", index=False)
```

```python
countries = pd.read_csv("countries.csv")

final = pd.merge(
    df,
    countries,
    on="Country",
    how="left"
)
```

```python
import pandas as pd

df = pd.read_csv("military_raw_data.csv")
```

```python
import pandas as pd
import re

def clean_number(x):
    if pd.isna(x):
        return x

    x = str(x)
    x = re.sub(r"[,%+]", "", x).strip()

    try:
        return float(x)
    except:
        return x

# Clean only text columns
for col in df.select_dtypes(include="object").columns:
    df[col] = df[col].map(clean_number)
```

```python
df = df.apply(lambda col: col.map(clean_number))
```

```python
df.to_csv("military_cleaned.csv", index=False)
print("Cleaning completed successfully!")
```
